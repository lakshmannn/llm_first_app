```python
import os
import subprocess
import ollama
import psutil
import shutil

MODEL_NAME = "gemma3:1b"

# ---------------------------------------------------------
# Check if Ollama service is running
# ---------------------------------------------------------
def check_ollama_running():
    try:
        subprocess.check_output(["ollama", "ps"])
        print("✔ Ollama service is running.")
        return True
    except Exception:
        print("✘ Ollama is NOT running.")
        print("  Start it manually using: ollama serve")
        return False

# ---------------------------------------------------------
# Check if Gemma3:1b model is installed locally
# ---------------------------------------------------------
def check_model_installed():
    try:
        models = ollama.list()
        names = [m["name"] for m in models["models"]]

        if MODEL_NAME in names:
            print(f"✔ Model '{MODEL_NAME}' is installed locally.")
            return True
        else:
            print(f"✘ Model '{MODEL_NAME}' is NOT installed.")
            print(f"  Install using: ollama pull {MODEL_NAME}")
            return False
    except Exception as e:
        print("✘ Could not list Ollama models.")
        print("  Error:", e)
        return False

# ---------------------------------------------------------
# Test model response
# ---------------------------------------------------------
def test_model_response():
    try:
        print("⏳ Testing model response...")
        response = ollama.chat(
            model=MODEL_NAME,
            messages=[{"role": "user", "content": "Hello Gemma, are you working?"}]
        )
        print("✔ Model responded successfully:")
        print("  →", response["message"]["content"])
        return True
    except Exception as e:
        print("✘ Model did NOT respond.")
        print("  Error:", e)
        return False

# ---------------------------------------------------------
# Check for cloud API keys (should NOT exist)
# ---------------------------------------------------------
def check_cloud_keys():
    keys = {
        "OPENAI_API_KEY": os.getenv("OPENAI_API_KEY"),
        "ANTHROPIC_API_KEY": os.getenv("ANTHROPIC_API_KEY"),
        "GOOGLE_API_KEY": os.getenv("GOOGLE_API_KEY"),
    }

    print("✔ Verifying network isolation (no cloud API keys)...")

    for name, value in keys.items():
        if value:
            print(f"⚠ {name} is set — but not needed for local Gemma3.")
        else:
            print(f"✔ {name} not set (correct for offline mode).")

# ---------------------------------------------------------
# Run all diagnostics
# ---------------------------------------------------------
def run_diagnostics():
    print("\n=== Local Gemma3:1b Diagnostic Report ===\n")

    check_ollama_running()
    check_model_installed()
    check_cloud_keys()
    test_model_response()

    print("\n=== Diagnostics Complete ===\n")


if __name__ == "__main__":
    run_diagnostics()
```
